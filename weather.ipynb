## Task 4: Create Timestamped Files

#### Use the touch command to create an empty file with a name in the YYYYmmdd_HHMMSS.txt format. You can achieve this by embedding your date command in backticks ` into the touch command. You should no longer use redirection (>>) in this step.

- Use the touch command with the date command embedded in backticks: touch `date +"%Y%m%d_%H%M%S"`.txt

 The touch command creates an empty file. By embedding the date command within backticks (`), the current date and time are formatted using the YYYYmmdd_HHMMSS pattern and used as the file name. For example, if the current date and time are November 14, 2026, at 1:00:03 PM, the command will create a file named 20261114_130003.txt.

### Explanation of the Data Set: 

The weather data set contains observations from the Athenry weather station provided by Met Eireann. The data includes various meteorological parameters such as temperature, humidity, wind speed, wind direction, and atmospheric pressure. Each observation is timestamped, providing a record of the weather conditions at specific times throughout the day.

According to data.gov.ie, the data set includes the following fields:

- temperature: The air temperature in degrees Celsius.
- humidity: The relative humidity as a percentage.
- wind_speed: The wind speed in meters per second.
- wind_direction: The wind direction in degrees.
- pressure: The atmospheric pressure in hectopascals (hPa).
- timestamp: The date and time of the observation.
